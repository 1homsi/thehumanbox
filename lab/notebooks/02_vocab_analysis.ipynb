# 02 — Vocab Analysis

Compare two lineage vocabularies using `language.vocab_diff` if available.

```python
try:
    from thehumanbox_lab.language.vocab_diff import vocab_diff, diff_summary
    HAS_VOCAB = True
except Exception as exc:
    HAS_VOCAB = False
    print(f'vocab_diff unavailable: {exc}')
```

```python
vocab_a = {'water': 'kuu', 'food': 'mef', 'danger': 'zix'}
vocab_b = {'water': 'kuu', 'food': 'paq', 'shelter': 'lon'}

if HAS_VOCAB:
    d = vocab_diff(vocab_a, vocab_b)
    print(diff_summary(d))
    print(d)
else:
    only_a = set(vocab_a) - set(vocab_b)
    only_b = set(vocab_b) - set(vocab_a)
    print(f'fallback: only_a={only_a} only_b={only_b}')
```
