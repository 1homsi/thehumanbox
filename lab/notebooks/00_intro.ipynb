# The Human Box — Lab Notebooks

This directory contains exploratory notebooks for analyzing simulation snapshots, vocab drift, and training data.

## Layout

- `src/thehumanbox_lab/viz/` — pure-stdlib SVG/HTML chart builders
- `src/thehumanbox_lab/language/` — vocab diff, dialect, edit distance, word frequency
- `scripts/build_report.py` — assembles `reports/latest.html`

## Conventions

- No external plotting deps. SVG is generated as strings.
- Notebooks fetch data from a running sim via `urllib` (no `requests`).
- Each notebook should be self-contained and runnable top-to-bottom.
