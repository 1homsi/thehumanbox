# 01 — Snapshot Explore

Fetch `/snapshot` from a running sim and print a summary.

```python
import json
import urllib.request

URL = 'http://localhost:8080/snapshot'

def fetch(url):
    try:
        with urllib.request.urlopen(url, timeout=3) as r:
            return json.loads(r.read().decode('utf-8'))
    except Exception as exc:
        return {'error': str(exc)}

snap = fetch(URL)
print(json.dumps({k: type(v).__name__ for k, v in snap.items()}, indent=2))
```

```python
orgs = snap.get('organisms', [])
print(f'organisms: {len(orgs)}')
lineages = {o.get('lineage_id') for o in orgs}
print(f'lineages: {len(lineages)}')
print(f'tick: {snap.get("tick")}')
```
